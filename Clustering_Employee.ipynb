B. Mengubah Kolom Pengalaman (experienceMas) ke Bulan (float)

```python
def process_exp(dataframe, column_name):
    dataframe[column_name] = dataframe[column_name].fillna('0 Year(s) 0 Month(s)')
    dataframe[column_name] = dataframe[column_name].replace('Fresher', '0 Year(s) 0 Month(s)')

    new_exp = []
    for val in dataframe[column_name]:
        try:
            parts = val.split(' ')
            year = 0
            month = 0

            if 'Year(s)' in parts:
                year_index = parts.index('Year(s)') - 1
                year = float(parts[year_index])
            if 'Month(s)' in parts:
                month_index = parts.index('Month(s)') - 1
                month = float(parts[month_index])

            total_month = year * 12 + month
            new_exp.append(total_month)
        except:
            new_exp.append(0)

    dataframe[column_name] = new_exp
    return dataframe

```

```python
df = process_exp(dataframe=df, column_name='experienceMas')
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

# Uji k untuk range tertentu, misal 1–10
for k in range(1, 20):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(df_final_2_normal)  # Gunakan data yang sudah distandardisasi
    inertia.append(kmeans.inertia_)

# Plot hasil
plt.plot(range(1, 20), inertia, marker='o')
plt.title('Elbow Method')
plt.xlabel('Jumlah Cluster (k)')
plt.ylabel('Inertia')
plt.grid(True)
plt.show()
```
